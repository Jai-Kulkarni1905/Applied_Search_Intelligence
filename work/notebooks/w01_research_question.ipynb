```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

# Load the dataset
df = pd.read_csv("content_refresh_anonymized.csv")

print(f"Total pages           : {len(df):,}")
print(f"Unique clients        : {df['client_id'].nunique()}")
print(f"Pages with 0 clicks   : {(df['clicks_90d'] == 0).sum():,} ({(df['clicks_90d'] == 0).mean()*100:.1f}%)")
print(f"Average impressions   : {df['impressions_90d'].mean():,.0f}")
print(f"Median impressions    : {df['impressions_90d'].median():,.0f}")
print(f"Average clicks        : {df['clicks_90d'].mean():.1f}")
print(f"Median clicks         : {df['clicks_90d'].median():.0f}")
```

**Inference**: The dataset shows a clear imbalance in content performance, with 44% of pages receiving zero clicks and the median page receiving just 1 click despite an average of 16.1 clicks. This indicates that a small number of pages account for most of the traffic, making it impractical for SEO teams to manually review every page. This is exactly the problem my project aims to solve by ranking pages based on their refresh opportunity, allowing content teams to focus their efforts on pages that are most likely to benefit from an update rather than relying on manual inspection or intuition.
